## Problem 10: Sherman–Morrison Inverse

---

Recall that the Sherman-Morrison formula says that, for any invertible $n \times n$ matrix $A$ and vectors $\vec u, \vec v \in \mathbb{R}^n$,

the inverse of

$$B = A + \vec u \vec v^T$$

is

$$B^{-1} = (A + \vec u \vec v^T)^{-1} = A^{-1} - \frac{A^{-1} \vec u \vec v^T A^{-1}}{1 + \vec v^T A^{-1} \vec u}$$

This requires $1+\vec v^T A^{-1}\vec u\ne0$ (and $A$ must be invertible).

Our goal here is to try and quantify how much "faster" using the Sherman-Morrison formula is, **once we already have $A^{-1}$**, than inverting $B = A + \vec u \vec v^T$ directly is.

To help us, we'll use Python's high-resolution `time.perf_counter` timer. See a demonstration of how it works below.


```python
x = time.perf_counter()
time.sleep(0.1)
y = time.perf_counter()
y - x
```

`y - x` is the approximate length of time, in seconds, that the code in between the two calls to `time.perf_counter()` took to run.

### Pixels and the RGB color model

An image is a grid of colored squares called **pixels**. In the **RGB** model, each pixel has three intensities: **red**, **green**, and **blue**, each ranging from 0 (none) to 255 (maximum).

We represent a pixel by a vector $\begin{bmatrix}R\\G\\B\end{bmatrix} \in \mathbb{R}^3$. For example:

| RGB values | Color |
|---|---|
| (255, 0, 0) | Red |
| (0, 255, 0) | Green |
| (0, 0, 255) | Blue |
| (0, 0, 0) | Black |
| (255, 255, 255) | White |
| (128, 128, 128) | Gray |

Run the embedded color picker below. Try changing one channel at a time, then set all three channels equal. What happens as their common value increases?

An **image filter** transforms each pixel's RGB vector into a new RGB vector. In this homework, color filters use a $3\times3$ matrix. The same matrix is applied to every pixel. Geometric transformations, introduced later, instead act on two-entry position vectors.


```python
# Run this supplied color explorer; no changes are needed.
import ipywidgets as widgets
from IPython.display import display

_rgb_picker = widgets.ColorPicker(value="#1e3c5a", description="Color")
_rgb_sliders = [widgets.IntSlider(value=v, min=0, max=255, description=k,
                                 continuous_update=True)
                for k, v in zip(["Red", "Green", "Blue"], [30, 60, 90])]
_rgb_preview = widgets.HTML()
_rgb_busy = False

def _rgb_update(change=None):
    global _rgb_busy
    if _rgb_busy:
        return
    _rgb_busy = True
    try:
        if change is not None and change["owner"] is _rgb_picker:
            values = tuple(int(_rgb_picker.value[i:i+2], 16) for i in (1, 3, 5))
            for slider, value in zip(_rgb_sliders, values):
                slider.value = value
        else:
            values = tuple(slider.value for slider in _rgb_sliders)
            _rgb_picker.value = "#{:02x}{:02x}{:02x}".format(*values)
        _rgb_preview.value = (
            f'<div style="height:90px;border:1px solid #888;border-radius:8px;'
            f'background:rgb{values};"></div><p><b>RGB: {values}</b></p>'
        )
    finally:
        _rgb_busy = False

_rgb_picker.observe(_rgb_update, names="value")
for _slider in _rgb_sliders:
    _slider.observe(_rgb_update, names="value")
_rgb_update()
display(widgets.VBox([_rgb_picker, *_rgb_sliders, _rgb_preview]))
```

## Grayscale (provided)

---
The grayscale filter replaces each component of a color vector, $\begin{bmatrix} R \\ G \\ B \end{bmatrix}$, by the average of all three components:

$$R_{\rm new}=G_{\rm new}=B_{\rm new}=\frac{R_{\rm old}+G_{\rm old}+B_{\rm old}}{3}$$

Each output uses the same three coefficients, so the matrix associated with this linear transformation is

$$G=\begin{bmatrix}1/3&1/3&1/3\\1/3&1/3&1/3\\1/3&1/3&1/3\end{bmatrix}$$

**We've implemented this filter for you.** `grayscale_matrix()` returns $G$, and `filter_pixel(A, u)` computes $A\vec u$. The first function chooses the matrix; the second applies it. Notice that `filter_pixel` takes the matrix as an argument, so we can reuse it for other color filters.

Read the two functions below and run the cell. `np.ones((3, 3))` creates a $3\times3$ array of ones; dividing by 3 divides every entry.

```python
def grayscale_matrix():
    """Average the original red, green, and blue values in every output row."""
    return np.ones((3, 3)) / 3

def filter_pixel(A, u):
    """Apply a color matrix A to one RGB column vector u."""
    return A @ u
```

## Stretching (provided)

So far, we've implemented color filters, which change a pixel's RGB vector. We'll now explore geometric transformations, which change a pixel's **position vector** $\begin{bmatrix}x\\y\end{bmatrix}$.

We use mathematical coordinates: $x$ increases rightward and $y$ increases upward, with the image centered at the origin. (This is different from how image coordinates are typically described, with $(0, 0)$ being the top-left of the image and $y$ increasing downwards.)

The supplied `scale_matrix(sx, sy)` makes sure that horizontal coordinates are multiplied by `sx` and vertical coordinates by `sy`. A factor of 1 leaves a coordinate unchanged; a factor between 0 and 1 compresses it.

The columns of a $2\times n$ matrix $X$ store position vectors. The supplied `transform_points(A, X)` computes $AX$ to transform all columns at once. Keep this helper unchanged.

Run the supplied functions and the following canvas cell and explore the two Stretch sliders.


```python
def transform_points(A, X):
    """Apply A to every position vector stored as a column of X."""
    return A @ X

def scale_matrix(sx, sy):
    """Returns a matrix that scales horizontal coordinates by sx and vertical coordinates by sy."""
    return np.array([[sx, 0], [0, sy]])
```
